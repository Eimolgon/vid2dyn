```python
import numpy as np
import sympy as sp
import sympy.physics.mechanics as me
```

```python
q1, q2, q3 = sp.symbols('q1, q2, q3')       # Angles
x_r, y_r, z_r = sp.symbols('x_r, y_r, z_r') # Position
r = sp.symbols('r')                         # Measures

N, R = sp.symbols('N, R', cls=me.ReferenceFrame)

R.orient_body_fixed(N, (q1, q2, q3), 'zxy')

O = me.Point('O')
Cr = me.Point('C_r')
w_t = me.Point('w_t') # wheel top
w_b = me.Point('w_b') # wheel bottom
w_e = me.Point('w_e') # wheel east
w_w = me.Point('w_w') # wheel west

O.set_pos(O, 0)
O.set_vel(N, 0)

Cr.set_pos(O, x_r*N.x + y_r*N.y + z_r*N.z)
w_t.set_pos(Cr, r*R.z)
w_b.set_pos(Cr, -r*R.z)
w_e.set_pos(Cr, r*R.x)
w_w.set_pos(Cr, -r*R.x)
```

```python
v1 = w_e.pos_from(O).express(N).dot(N.x)

eval_v1 = sp.lambdify((q1, q2, q3, x_r, y_r, z_r, r), v1)

sub0 = (0, 0, 0, 10, 0, 10, 5)
sub45 = (np.deg2rad(45), 0, 0, 10, 0, 10, 5)
sub90 = (np.deg2rad(90), 0, 0, 10, 0 ,10, 5)


f0 = eval_v1(*sub0)
f45 = eval_v1(*sub45)
f90 = eval_v1(*sub90)
```
