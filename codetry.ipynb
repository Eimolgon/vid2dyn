```python
import numpy as np
import sympy as sp
import sympy.physics.mechanics as me
```

```python
q1, q2, q3 = sp.symbols('q1, q2, q3')               # Angles, q1 yaw q2 roll
x_r, y_r, z_r = sp.symbols('x_r, y_r, z_r') # Position
r = sp.symbols('r')                         # Measures

N, R = sp.symbols('N, R', cls=me.ReferenceFrame)

R.orient_body_fixed(N, (q1, q2, q3), 'zxy')

#Seen from the right
O = me.Point('O')
Cr = me.Point('C_r')
p1 = me.Point('w_t') # top
p2 = me.Point('w_f') # front
p3 = me.Point('w_b') # bottom
p4 = me.Point('w_r') # rear

O.set_pos(O, 0)
O.set_vel(N, 0)

Cr.set_pos(O, x_r*N.x + y_r*N.y + z_r*N.z)
p1.set_pos(Cr, r*R.z)
p2.set_pos(Cr, r*R.x)
p3.set_pos(Cr, -r*R.z)
p4.set_pos(Cr, -r*R.x)
```

```python
p1.pos_from(Cr).dot(N.z)
```

```python
v1 = p2.pos_from(O).express(N).dot(N.x)

eval_v1 = sp.lambdify((q1, q2, x_r, y_r, z_r, r), v1)

sub0 = (0, 0, 10, 0, 10, 5)
sub45 = (np.deg2rad(45), 0, 10, 0, 10, 5)
sub90 = (np.deg2rad(90), 0, 10, 0 ,10, 5)


f0 = eval_v1(*sub0)
f45 = eval_v1(*sub45)
f90 = eval_v1(*sub90)
```
