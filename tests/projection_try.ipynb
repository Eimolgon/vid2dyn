```python
import numpy as np
import sympy as sp
import sympy.physics.mechanics as me
```

```python
q1, q2, q3, q4, q5, q6 = sp.symbols('q1, q2, q3, q4, q5, q6')
x1, y1, z1, x2, y2, z2 = sp.symbols('x1, y1, z1, x2, y2, z2')
l = sp.symbols('l')

N, A, B = sp.symbols('N, A, B', cls=me.ReferenceFrame)

A.orient_body_fixed(N, (q1, q2, q3), 'xyz')
B.orient_body_fixed(N, (q4, q5, q6), 'xyz')

O = me.Point('O')
P = me.Point('P')
s1 = me.Point('s1')
s2 = me.Point('s2')
s3 = me.Point('s3')
s4 = me.Point('s4')

O.set_pos(O, 0)
O.set_vel(N, 0)

# Pinhole position
P.set_pos(O, x1*N.x + y1*N.y + z1*N.z)

# Bottom left corner position
s1.set_pos(O, x2*N.x + y2*N.y + z2*N.z)

s2.set_pos(s1, l*B.x)
s3.set_pos(s2, l*B.z)
s4.set_pos(s1, l*B.z)
```
