```python
import numpy as np
import sympy as sp
import sympy.physics.mechanics as me
import matplotlib.pyplot as plt
```

```python
q1, q2, q3, q4, q5, q6, q7, q8, q9 = \
    sp.symbols('q1, q2, q3, q4, q5, q6, q7, q8, q9')
x1, y1, z1, x2, y2, z2 = \
    sp.symbols('x1, y1, z1, x2, y2, z2')
l, f, wi, hi = \
    sp.symbols('l, f, w_i, h_i')

# Being N the world frame
N, A, B, C = sp.symbols('N, A, B, C', cls=me.ReferenceFrame)

C.orient_body_fixed(N, (q1, q2, q3), 'xyz') # Camera frame
A.orient_body_fixed(N, (q4, q5, q6), 'xyz') # Body 1 frame
B.orient_body_fixed(A, (q7, q8, q9), 'xyz') # Body 2 frame

O = me.Point('O')
P = me.Point('P')

s1 = me.Point('s1')
s2 = me.Point('s2')
s3 = me.Point('s3')
s4 = me.Point('s4')

t1 = me.Point('t1')
t2 = me.Point('t2')
t3 = me.Point('t3')
t4 = me.Point('t4')

O.set_pos(O, 0)
O.set_vel(N, 0)

# Pinhole position
P.set_pos(O, x1*N.x + y1*N.y + z1*N.z)

# Bottom left corner position
s1.set_pos(O, x2*N.x + y2*N.y + z2*N.z)

s2.set_pos(s1, l*A.x)
s3.set_pos(s2, l*A.z)
s4.set_pos(s1, l*A.z)

t1.set_pos(s1, 5*x1*N.x + 10 - y1*N.y + z1*N.z)
```

```python
cameraIntrinsics = {'f': 28, 'wi': 1920, 'hi': 1080}
cameraExtrinsics = {'x1': 10, 'y1': 10, 'z1': 0, 'q1': 0, 'q2': 0, 'q3': -np.pi/2}
# subs = {'l': 5}

body_1_params = {'q4': 0, 'q5': 0, 'q6': np.pi/4}
body_2_params = {'q4': 0, 'q5': 0, 'q6': np.pi/2}

# Considering z as the vertical axis of the screen, and y the depth from the 
# camera

r_s1_P = s1.pos_from(P).express(C)
x_c = r_s1_P.dot(C.x)
y_c = r_s1_P.dot(C.y)
z_c = r_s1_P.dot(C.z)


x_i = f * x_c / y_c
z_i = f * z_c / y_c 
```

```python
subs = {
    f: cameraIntrinsics['f'],
    x1: cameraExtrinsics['x1'],
    y1: cameraExtrinsics['y1'],
    z1: cameraExtrinsics['z1'],
    q1: cameraExtrinsics['q1'],
    q2: cameraExtrinsics['q2'],
    q3: cameraExtrinsics['q3'],
    l: 5
}

x_i_num = x_i.subs(subs)
z_i_num = z_i.subs(subs)
```

```python
plt.plot(x_i_num, z_i_num)
plt.show()
```
