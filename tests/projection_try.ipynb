```python
import numpy as np
import sympy as sp
import sympy.physics.mechanics as me
```

```python
q1, q2, q3, q4, q5, q6, q7, q8, q9 = sp.symbols('q1, q2, q3, q4, q5, q6, q7, q8, q9')
x1, y1, z1, x2, y2, z2 = sp.symbols('x1, y1, z1, x2, y2, z2')
l, f, wi, hi = sp.symbols('l, f, w_i, h_i')

# Being N the world frame
N, A, B, C = sp.symbols('N, A, B, C', cls=me.ReferenceFrame)

C.orient_body_fixed(N, (q1, q2, q3), 'xyz') # Camera frame
A.orient_body_fixed(N, (q4, q5, q6), 'xyz') # Body 1 frame
B.orient_body_fixed(A, (q7, q8, q9), 'xyz') # Body 2 frame

O = me.Point('O')
P = me.Point('P')

s1 = me.Point('s1')
s2 = me.Point('s2')
s3 = me.Point('s3')
s4 = me.Point('s4')

O.set_pos(O, 0)
O.set_vel(N, 0)

# Pinhole position
P.set_pos(O, x1*N.x + y1*N.y + z1*N.z)

# Bottom left corner position
s1.set_pos(O, x2*N.x + y2*N.y + z2*N.z)

s2.set_pos(s1, l*A.x)
s3.set_pos(s2, l*A.z)
s4.set_pos(s1, l*A.z)
```
