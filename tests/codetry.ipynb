```python
import numpy as np
import sympy as sp
import sympy.physics.mechanics as me
```

```python
q1, q2, q3 = sp.symbols('q1, q2, q3')               # Angles, q1 yaw q2 roll
x_r, y_r, z_r = sp.symbols('x_r, y_r, z_r') # Position
r = sp.symbols('r')                         # Measures

N, R = sp.symbols('N, R', cls=me.ReferenceFrame)

R.orient_body_fixed(N, (q1, q2, q3), 'zxy')

#Seen from the right
O = me.Point('O')
Cr = me.Point('C_r')
p1 = me.Point('w_t') # top
p2 = me.Point('w_f') # front
p3 = me.Point('w_b') # bottom
p4 = me.Point('w_r') # rear

O.set_pos(O, 0)
O.set_vel(N, 0)

Cr.set_pos(O, x_r*N.x + y_r*N.y + z_r*N.z)
p1.set_pos(Cr, r*R.z)
p2.set_pos(Cr, r*R.x)
p3.set_pos(Cr, -r*R.z)
p4.set_pos(Cr, -r*R.x)
```

```python
def find_points(points):

    bottom_point_idx = points.index(min(points))

    if bottom_point_idx == 0:
        bottom_point = points[0]
        upper_point = points[2]
    elif bottom_point_idx == 1:
        bottom_point = points[1]
        upper_point = points[3]
    elif bottom_point_idx == 2:
        bottom_point = points[2]
        upper_point = points[0]
    else:
        bottom_point = points[3]
        upper_point = points[1]


    return upper_point, bottom_point
```
